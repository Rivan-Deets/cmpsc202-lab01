```python
import matplotlib.pyplot as plt
import numpy as np
from time import perf_counter


def helper(length):
    """"helper function used to generate lists """
    length = int(length)
    if length < 2:
        raise ValueError("length must be at least 2")

    values = np.random.randint(-100, 101, size=length)
    if np.all(values >= 0):
        values[0] = -100
    elif np.all(values <= 0):
        values[0] = 100

    return values.tolist()


def baseline_max_sublist(arr):
    best_sum = arr[0]
    best_start = 0
    best_end = 0

    for start in range(len(arr)): # gets length of the array and makes each item a start point
        current_sum = 0
        for end in range(start, len(arr)): #doesthe sae but for the ed point of the array
            current_sum += arr[end] 
            if current_sum > best_sum: #if current sum is bigger then it becomes the best sum
                best_sum = current_sum # rest of if statement saves the othervaribles of the newest best sum
                best_start = start
                best_end = end

    return best_sum, best_start, best_end


def kadane_max_sublist(arr):
    current = arr[0]
    best = arr[0]
    best_start = 0
    best_end = 0
    current_start = 0

    for index, num in enumerate(arr[1:], start=1): #runs through the array
        if num > current + num: # saves the current number 
            current = num
            current_start = index
        else:
            current += num

        if current > best: # updatesthe best number with the current 
            best = current
            best_start = current_start
            best_end = index

    return best, best_start, best_end


def benchmark_algorithms(): #runs the list lengths of a higher quantity
    lengths = [100, 500, 1000, 2500, 5000, 10000]
    baseline_times = []
    kadane_times = []

    print(f"{'N':>6} {'Baseline (s)':>15} {'Kadane (s)':>15}")
    print("-" * 40)

    for length in lengths:
        data = helper(length)

        start_time = perf_counter() #saves the counting time for the basline sublist
        baseline_result = baseline_max_sublist(data)
        baseline_time = perf_counter() - start_time
        baseline_times.append(baseline_time)

        start_time = perf_counter() #same but for the kadane sublist
        kadane_result = kadane_max_sublist(data)
        kadane_time = perf_counter() - start_time
        kadane_times.append(kadane_time)

        assert baseline_result[0] == kadane_result[0]
        print(f"{length:6d} {baseline_time:15.6f} {kadane_time:15.6f}")

    plt.figure(figsize=(9, 6)) #plots the arrays for both sublists
    plt.plot(lengths, baseline_times, marker="o", linewidth=2, label="Baseline")
    plt.plot(lengths, kadane_times, marker="s", linewidth=2, label="Kadane")
    plt.yscale("log")
    plt.title("Maximum Sublist Algorithm Performance")
    plt.xlabel("Array Size (n)")
    plt.ylabel("Execution Time (seconds)")
    plt.xticks(lengths)
    plt.grid(True, which="both", linestyle="--", alpha=0.5)
    plt.legend()
    plt.tight_layout()
    plt.show()


benchmark_algorithms()
```
